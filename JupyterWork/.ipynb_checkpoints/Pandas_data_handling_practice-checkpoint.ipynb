## Fixing issues

```python
dataframe = df.copy()

dataframe = dataframe.assign(release_date=pd.to_datetime(dataframe["release_date"]))

dataframe = dataframe.drop(columns = ["origin_country"])

dataframe["release_year"] = pd.to_numeric(dataframe["release_year"]).astype("Int64")

dataframe = dataframe.dropna()

count = df["id"].count()

cleaned_count = dataframe["id"].count()

# dataframe.info()
```

```python
dataframe.head()
```

```python
tv_shows = dataframe[dataframe["media_type"] == "tv"].count()[0]
movies = dataframe[dataframe["media_type"] == "movie"].count()[0]
```

```python
print(f"In the cleaning process we removed {count - cleaned_count} rows and 1 column.")
```

```python
print(f"In the cleaned dataset there is {(tv_shows / (tv_shows + movies) * 100):.2f}% tv shows and {(movies / (tv_shows + movies) * 100):.2f}% movies.")
```

```python
unique_genres = dataframe.groupby("genres")["id"].sum().count()

print(f"There are {unique_genres} unique genres in the cleaned dataset.")
```

```python
unique_languages = dataframe.groupby("original_language")["id"].sum().count()

print(f"There are {unique_languages} unique languages in the cleaned dataset.")
```

```python
average_rating = dataframe["vote_average"].mean()

print(f"The average rating over the cleaned set is {average_rating:.2f}")
```

```python
average_popularity = dataframe["popularity"].mean()
highest_popularity = dataframe["popularity"].max()
lowest_popularity = dataframe["popularity"].min()


print(f"The average attendence in the cleaned dataset is {average_popularity:.2f}, with the highest being {highest_popularity:.2f} and the lowest being {lowest_popularity:.2f}")
```
