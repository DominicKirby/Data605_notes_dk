```python
import json
import pandas as pd
```

```python
with open("final_data_new.json") as f:
    df = pd.DataFrame(json.load(f))
```

```python
df.tail()
```

```python
amount_null = df.isnull().sum()
amount_null.sum()
amount_null
```

```python
df.columns
```

```python
cleaned_df = df.drop(columns = ["publisher", "pegi"])
cleaned_df = cleaned_df.dropna()
```

```python
cleaned_df.info()
```

```python
cleaned_df = cleaned_df.replace("free", 0)
```

```python
cleaned_df["price"] = pd.to_numeric(cleaned_df["price"]) / 10
```

```python
cleaned_df.info()
```

```python
average_cost = cleaned_df["price"].mean()

max_cost = cleaned_df["price"].max()

min_cost = cleaned_df["price"].min()

print(f"The average cost of a game on steam is {average_cost}, with max cost of {max_cost} and minimum cost of {min_cost}")
```
