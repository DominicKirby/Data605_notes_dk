# Seaborn vs MatPlotLib

```python
# %pip install matplotlib
# %pip install seaborn

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
fmri = sns.load_dataset('fmri')

fmri.head()
```

```python
sns.lineplot(data = fmri, x = 'timepoint', y = 'signal')
plt.title("Average signal over timepoints with CI at 95%")
plt.xlabel('Timepoint')
plt.ylabel('Signal')
```

```python
fig, ax = plt.subplots()

ax.plot('timepoint', 'signal', data = fmri)
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.barplot(data = tips, x = 'day', y ='total_bill', hue = "time")
plt.title("Average bill cost per day of the week open")
plt.xlabel("Day")
plt.ylabel("Total Bill")
```

```python
plt.bar(tips.day, tips.total_bill)
```

```python
flights = sns.load_dataset('flights')

# LONG FORMAT

flights.head()
```

```python
# Pivoting the data
# WIDE FORMAT

fp = flights.pivot(index="year", columns = "month", values = "passengers")
```

```python
fp.head()
```

```python
sns.heatmap(fp)
```

## Dataset practice

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
sns.scatterplot(data = iris, x = "petal_length", y = "petal_width", hue = "species")
plt.title("Petal length vs Width for each species of plant")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
```

```python
plt.scatter(data = iris[iris["species"] == 'setosa'], x = "petal_length", y = "petal_width", label = "setosa", s = 10)
plt.scatter(data = iris[iris["species"] == 'versicolor'], x = "petal_length", y = "petal_width", label = "versicolor", s = 10)
plt.scatter(data = iris[iris["species"] == 'virginica'], x = "petal_length", y = "petal_width", label = "virginica", s = 10)
plt.title("Petal Length vs Width for each species of plant")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.show()
```

```python
sns.scatterplot(data = iris, x = "sepal_length", y = "sepal_width", hue = "species")
plt.title("Sepal Length vs Width for each species of plant")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
```

```python
plt.scatter(data = iris[iris["species"] == 'setosa'], x = "sepal_length", y = "sepal_width", label = "setosa")
plt.scatter(data = iris[iris["species"] == 'versicolor'], x = "sepal_length", y = "sepal_width", label = "versicolor")
plt.scatter(data = iris[iris["species"] == 'virginica'], x = "sepal_length", y = "sepal_width", label = "virginica")
plt.title("Sepal Length vs Width for each species of plant")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.show()
```
